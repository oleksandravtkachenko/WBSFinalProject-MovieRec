```python
# Personality quiz — 3 questions to determine user's genre preferences
# Each answer maps to a list of genres that get counted
# The most frequently selected genres become the user's top preferences

QUIZ = [
    {"q": "Friday night. What are you doing?",
     "opts": [("Hosting a dinner party", ["Comedy","Romance"]), # social: Comedy, Romance
              ("Watching the news anxiously", ["Thriller","Drama"]), # tense: Thriller, Drama
              ("Planning your next adventure", ["Adventure","Sci-Fi"]), # explorer: Adventure, Sci-Fi
              ("Reading or journaling", ["Drama","Documentary"])]}, # reflective: Drama, Documentary
    {"q": "Pick a villain",
     "opts": [("The Joker — chaotic genius", ["Thriller","Action"]), # chaos: Thriller, Action
              ("Nurse Ratched — cold control", ["Drama","Horror"]), # control: Drama, Horror
              ("Thanos — tragic idealist", ["Sci-Fi","Action"]), # idealist: Sci-Fi, Action
              ("Amy Dunne — calculated", ["Thriller","Drama"])]}, # calculated: Thriller, Drama
    {"q": "Your ideal movie ending?",
     "opts": [("Everyone lives happily", ["Comedy","Romance"]), # optimist: Comedy, Romance
              ("Bittersweet and realistic", ["Drama"]), # realist: Drama
              ("Shocking twist", ["Thriller","Sci-Fi"]), # thrill-seeker: Thriller, Sci-Fi
              ("Left open to interpretation", ["Drama","Documentary"])]} # thinker: Drama, Documentary
]

def quiz_to_genres(answers: list) -> list:
    """
    Convert quiz answers into top 3 genre recommendations.
    Each answer contains a list of genres - count all genres across answers.
    Returns the 3 most frequently appearing genres sorted by count.
    Example: [["Comedy","Romance"], ["Thriller"], ["Comedy"]]: ["Comedy","Romance","Thriller"]
    """
    counts = {}
    for genres in answers:
        for g in genres:
            counts[g] = counts.get(g, 0) + 1 # count each genre occurrence
    return sorted(counts, key=counts.get, reverse=True)[:3] # return top 3
```

```python
# Simulate user choosing the 1st option for all 3 questions
test_answers = [
    ["Comedy", "Romance"],  # From Q1
    ["Thriller", "Action"],  # From Q2
    ["Comedy", "Romance"]   # From Q3
]

top_genres = quiz_to_genres(test_answers)
print(f"Top 3 Genres: {top_genres}")
# Expected output: ['Comedy', 'Romance', 'Thriller'] (or 'Action')
```

```python
import unittest

class TestMovieMatchmaker(unittest.TestCase):
    def test_basic_scoring(self):
        # Case: Clear winners
        answers = [["Action", "Sci-Fi"], ["Action"], ["Sci-Fi", "Drama"], ["Action"]]
        # Action=3, Sci-Fi=2, Drama=1
        self.assertEqual(quiz_to_genres(answers), ["Action", "Sci-Fi", "Drama"])

    def test_tie_handling(self):
        # Case: Tie for the 3rd spot
        answers = [["Comedy"], ["Drama"], ["Horror"]]
        result = quiz_to_genres(answers)
        self.assertEqual(len(result), 3)
        self.assertIn("Comedy", result)

    def test_few_genres(self):
        # Case: Less than 3 unique genres provided
        answers = [["Action"], ["Action"]]
        self.assertEqual(quiz_to_genres(answers), ["Action"])

# Run the test
unittest.main(argv=[''], exit=False)
```
