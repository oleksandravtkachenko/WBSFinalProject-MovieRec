```python
from dotenv import load_dotenv
import os
from groq import Groq

# Load API keys from .env file
load_dotenv()

# Initialize Groq client with API key
client = Groq(api_key=os.getenv("GROQ_API_KEY"))

def detect_mood(text: str) -> str:
    """
    Detect the mood from free text input using Groq LLM.
    Classifies into exactly one of five moods:
    Happy, Tense, Romantic, Adventurous, Thoughtful.

    Example:
    detect_mood("I had a stressful day"): "Tense"
    detect_mood("Let's watch something fun!"): "Happy"
    """
    r = client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=[{"role":"user","content":
            f'Classify into exactly one: Happy, Tense, Romantic, Adventurous, Thoughtful.\n'
            f'Text: "{text}"\nReturn only the mood word.'}] # strict prompt — one word only
    )
    return r.choices[0].message.content.strip() # remove whitespace from response
```

```python
# Assuming detect_mood is defined in a cell above
# Test 1 - romantic mood
test_input = "We want a relaxing evening watching something romantic."
print(f"Detected Mood: {detect_mood(test_input)}")
```

```python
# Test 2 — tense mood
result = detect_mood("I had a stressful day and want something intense and thrilling.")
print(f"Test 2 — Got: {result}")
print(f"Valid mood returned: {result in ['Happy', 'Tense', 'Romantic', 'Adventurous', 'Thoughtful']}")
print()
```

```python
# Test 3 — happy mood
result = detect_mood("I feel great today, let's watch something fun and cheerful!")
print(f"Test 3 — Expected: Happy — Got: {result}")
print(f"Correct: {result == 'Happy'}")
print()
```

```python
# Test 4 — returns only one word
result = detect_mood("I want to explore something new and exciting")
print(f"Test 4 — Returns single word: {len(result.split()) == 1}")
print(f"Valid mood: {result in ['Happy', 'Tense', 'Romantic', 'Adventurous', 'Thoughtful']}")
```
