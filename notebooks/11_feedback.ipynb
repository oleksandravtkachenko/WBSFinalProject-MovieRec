```python
import pandas as pd
from datetime import datetime
DATA_DIR = "../data"

def save_preferences(user_id, mood, genres, favorite, avoid):
    pd.DataFrame([{"user_id":user_id,"mood":mood,"genres":str(genres),
                   "favorite":favorite,"avoid":str(avoid),
                   "ts":datetime.now().isoformat()}]).to_csv(
        f"{DATA_DIR}/user_preferences.csv", mode="a", header=False, index=False)

def load_last_preferences(user_id):
    try:
        df = pd.read_csv(f"{DATA_DIR}/user_preferences.csv",
            names=["user_id","mood","genres","favorite","avoid","ts"])
        u = df[df.user_id == user_id]
        return u.iloc[-1] if not u.empty else None
    except: return None

def save_feedback(user_id, title, feedback):
    pd.DataFrame([[user_id, title, feedback, datetime.now()]]).to_csv(
        f"{DATA_DIR}/feedback.csv", mode="a", header=False, index=False)

def save_watchlist(user_id, title, year):
    pd.DataFrame([[user_id, title, year, datetime.now()]]).to_csv(
        f"{DATA_DIR}/watchlist.csv", mode="a", header=False, index=False)

def get_disliked(user_id):
    try:
        df = pd.read_csv(f"{DATA_DIR}/feedback.csv",
            names=["user_id","title","feedback","ts"])
        return df[(df.user_id==user_id)&(df.feedback==-1)]["title"].tolist()
    except: return []
```

```python
# Test 1 — save and load preferences
save_preferences("alice", "Happy", ["Comedy", "Romance"], "Notting Hill", ["Horror"])
prefs = load_last_preferences("alice")
print("Test 1 — Save and load preferences:")
print(f"User: {prefs['user_id']}")
print(f"Mood: {prefs['mood']}")
print(f"Favorite: {prefs['favorite']}")
print(f"Loaded correctly: {prefs['user_id'] == 'alice' and prefs['mood'] == 'Happy'}")
print()
```

```python
# Test 2 — save feedback and get disliked
save_feedback("alice", "Transformers", -1)
save_feedback("alice", "Inception", 1)
disliked = get_disliked("alice")
print("Test 2 — Feedback and disliked:")
print(f"Disliked movies: {disliked}")
print(f"Transformers in disliked: {'Transformers' in disliked}")
print(f"Inception not in disliked: {'Inception' not in disliked}")
print()
```

```python
# Test 3 — save watchlist
save_watchlist("alice", "Inception", 2010)
df = pd.read_csv(f"{DATA_DIR}/watchlist.csv", names=["user_id","title","year","ts"])
alice_watchlist = df[df.user_id == "alice"]["title"].tolist()
print("Test 3 — Watchlist:")
print(f"Watchlist: {alice_watchlist}")
print(f"Inception in watchlist: {'Inception' in alice_watchlist}")
```
