```python
import pandas as pd
import random
from datetime import datetime, timedelta
import os
# Automatically detect correct data folder path
# Works from both root folder and notebooks/ subfolder
if os.path.exists("data"):
    DATA_DIR = "data" # running from root
else:
    DATA_DIR = "../data" # running from notebooks/

# Simulated user profiles with predefined moods, genres and favorite movies
# These represent realistic user personas for testing the recommendation pipeline
USERS = [
    {"user_id":"hanna","mood":"Happy","genres":["Comedy","Animation"],"favorite":"The Grand Budapest Hotel"},
    {"user_id":"christoph","mood":"Tense","genres":["Thriller","Action"], "favorite":"Heat"},
    {"user_id":"barbara","mood":"Romantic","genres":["Romance","Drama"],   "favorite":"Notting Hill"},
    {"user_id":"thomas","mood":"Adventurous","genres":["Sci-Fi","Action"],  "favorite":"Interstellar"},
    {"user_id":"oleksandra","mood":"Thoughtful","genres":["Drama","Documentary"],"favorite":"Parasite"},
]
# Pool of movies used to generate simulated feedback history
MOVIES = ["Inception","WALL-E","The Martian","Arrival","About Time",
          "Hitch","Parasite","Her","The Dark Knight","Amélie"]

def generate():
    """
    Generate simulated feedback and preferences for all users.
    Each user randomly rates 5 movies from the pool.
    Feedback is weighted 3:1 positive to negative to simulate realistic behavior.
    Saves results to feedback.csv and user_preferences.csv.
    """
    records = []
    for u in USERS:
        for m in random.sample(MOVIES, 5):  # pick 5 random movies per user
            records.append({"user_id":u["user_id"],"title":m,
                "feedback":random.choice([1,1,1,-1]), # 75% positive, 25% negative
                "ts": datetime.now()-timedelta(days=random.randint(1,30))}) # random date in last 30 days
    pd.DataFrame(records).to_csv(f"{DATA_DIR}/feedback.csv", index=False)
    pd.DataFrame(USERS).to_csv(f"{DATA_DIR}/user_preferences.csv", index=False)

generate()
print("Simulated users generated")
```

```python
# Test 1 — generate runs without errors
generate()
print("Test 1 — Generate completed without errors")
print()
```

```python
# Test 2 — check feedback.csv has correct structure
df_feedback = pd.read_csv(f"{DATA_DIR}/feedback.csv")
print("Test 2 — Feedback CSV:")
print(f"Total records: {len(df_feedback)} (expected 25 — 5 users x 5 movies)")
print(f"Columns correct: {list(df_feedback.columns) == ['user_id','title','feedback','ts']}")
print(f"All users present: {set(df_feedback.user_id.unique()) == {'hanna','christoph','barbara','thomas','oleksandra'}}")
print()
```

```python
# Test 3 — check user_preferences.csv
df_prefs = pd.read_csv(f"{DATA_DIR}/user_preferences.csv")
print("Test 3 — Preferences CSV:")
print(f"Total users: {len(df_prefs)} (expected 5)")
print(f"All users present: {set(df_prefs.user_id.unique()) == {'hanna','christoph','barbara','thomas','oleksandra'}}")
print(f"Feedback values valid: {set(df_feedback.feedback.unique()).issubset({1,-1})}")
```
